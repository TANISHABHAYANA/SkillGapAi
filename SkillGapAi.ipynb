```python
import pandas as pd

df = pd.read_excel("Career Dataset.xlsx")

# Column name fix (agar needed ho)
df = df.rename(columns={
    'Career': 'Recommended_Career'
})

# Unique careers
careers = df['Recommended_Career'].dropna().unique()

print("Total unique careers:", len(careers))
print("\nList of careers:\n")

for i, c in enumerate(sorted(careers), 1):
    print(f"{i}. {c}")
```

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    df['Input'], df['Recommended_Career'],
    test_size=0.2,
    random_state=42,
    stratify=df['Recommended_Career']
)
```

```python
#from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    ngram_range=(1,3),
    max_df=0.85,
    min_df=1,
    stop_words='english'
)

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)
```

```python
dt = DecisionTreeClassifier()
lr = LogisticRegression(max_iter=1000, class_weight='balanced')
nb = MultinomialNB()
rf = RandomForestClassifier(n_estimators=200,random_state=42)
sgd = SGDClassifier(loss='log_loss',class_weight='balanced')
knn = KNeighborsClassifier(n_neighbors=5)
#hgb = HistGradientBoostingClassifier()
ridge = RidgeClassifier()

dt.fit(X_train, y_train)
lr.fit(X_train, y_train)
nb.fit(X_train, y_train)
rf.fit(X_train, y_train)
sgd.fit(X_train, y_train)
knn.fit(X_train, y_train)
ridge.fit(X_train,y_train)
#hgb.fit(X_train.toarray(),y_train)
```

```python
y_pred_dt = dt.predict(X_test)
y_pred_lr = lr.predict(X_test)
y_pred_nb = nb.predict(X_test)
y_pred_rf = rf.predict(X_test)
y_pred_sgd = sgd.predict(X_test)
y_pred_knn = knn.predict(X_test)
y_pred_ridge = ridge.predict(X_test)
#y_pred_hgb = hgb.predict(X_test.toarray())
```

```python
def evaluate_model(cm):
  total = np.sum(cm)
  correct = np.trace(cm)
  accuracy = correct/total

  precision_vals = []
  recall_vals = []

  for i in range(len(cm)):
    TP = cm[i][i]
    FP = sum(cm[:,i]) - TP
    FN = sum(cm[i,:]) - TP

    precision_vals.append(0 if (TP+FP) == 0 else TP/(TP+FP))
    recall_vals.append(0 if(TP+FN) == 0 else TP/(TP+FN))

    precision = np.mean(precision_vals)
    recall = np.mean(recall_vals)

    f1 = 0 if(precision + recall) == 0 else 2* (precision * recall) / (precision + recall)
    return accuracy,precision,recall,f1
```

```python

cm_dt = confusion_matrix(y_test, y_pred_dt)
acc_dt,pre_dt,rec_dt,f1_dt = evaluate_model(cm_dt)
plt.figure(figsize=(6,5))
sns.heatmap(cm_dt, annot=True, fmt='d', cmap='Blues')

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - Decision Tree")

plt.show()

print("\nDecision Tree:")
print("Accuracy:",acc_dt)
print("Precision:",pre_dt)
print("Recall:",rec_dt)
print("F1 Score:",f1_dt)
```

```python
cm_lr = confusion_matrix(y_test,y_pred_lr)
acc_lr,pre_lr,rec_lr,f1_lr = evaluate_model(cm_lr)


plt.figure(figsize=(6,5))
sns.heatmap(cm_lr, annot=True, fmt='d', cmap='Blues')

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - Logistic Regression")

plt.show()
print("\nLogistic Regression:")
print("Accuracy:",acc_lr)
print("Precision:",pre_lr)
print("Recall:",rec_lr)
print("F1 Score:",f1_lr)
```
